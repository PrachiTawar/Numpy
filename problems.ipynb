```python
#Inserting elements in rows and columns of 2d array.
import numpy as np
arr=np.arange(12).reshape(4,3)
print('\n\n2D arr:\n',arr)
print('\n\nShape:',arr.shape)
arr=np.insert(arr,1,5,axis=1)
print('\n\n Array After insertion:',arr)
print('\n\nShape:',arr.shape)
arr=np.insert(arr,2,6,axis=0)
print('\n\n Array After insertion:',arr)
print('\n\nShape:',arr.shape)
```
