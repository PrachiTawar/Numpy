```python
# Eye Function that is identity fucntion where diagonal are 1 and rest other values are 0 in matrix.
import numpy as np
# Create a 3x3 identity matrix
identity_matrix = np.eye(3)
print("3x3 Identity Matrix:")
print(identity_matrix)

# Create a 4x5 matrix with ones on the main diagonal
matrix_main_diag = np.eye(4, 5)
print("\n4x5 Matrix with ones on the main diagonal:")
print(matrix_main_diag)

# Create a 4x4 matrix with ones on the first upper diagonal (k=1)
matrix_upper_diag = np.eye(4, k=1)
print("\n4x4 Matrix with ones on the first upper diagonal:")
print(matrix_upper_diag)

# Create a 4x4 matrix with ones on the first lower diagonal (k=-1)
matrix_lower_diag = np.eye(4, k=-1)
print("\n4x4 Matrix with ones on the first lower diagonal:")
print(matrix_lower_diag)
```

```python
#Linespace function 
import numpy as np
# Create an array of 10 evenly spaced numbers between 0 and 1
linspace_array = np.linspace(0, 1, 10)
print("Array of 10 evenly spaced numbers between 0 and 1:")
print(linspace_array)

# Create an array of 5 evenly spaced numbers between -5 and 5, excluding the endpoint
linspace_no_endpoint = np.linspace(-5, 5, 5, endpoint=False)
print("\nArray of 5 evenly spaced numbers between -5 and 5 (excluding endpoint):")
print(linspace_no_endpoint)

# Create an array and get the step size
linspace_with_step, step_size = np.linspace(0, 10, 5, retstep=True)
print("\nArray with step size:")
print(linspace_with_step)
print("Step size:", step_size)
```

```python
#Reshape function with differnt dimensional Array
import numpy as np
# Create a 1D array
arr = np.array([1, 2, 3, 4, 5, 6])
print("Original array:")
print(arr)
print("Shape:", arr.shape)

# Reshape the 1D array into a 2x3 array
reshaped_arr = np.reshape(arr, (2, 3))
print("\nReshaped array (2x3):")
print(reshaped_arr)
print("Shape:", reshaped_arr.shape)

# Reshape the 1D array into a 3x2 array using -1
reshaped_arr_auto = np.reshape(arr, (3, -1))
print("\nReshaped array (3x2 using -1):")
print(reshaped_arr_auto)
print("Shape:", reshaped_arr_auto.shape)

# Reshape the 1D array into a 2x3 array and modify an element
reshaped_arr_modify = np.reshape(arr, (2, 3))
reshaped_arr_modify[0, 0] = 100
print("\nReshaped array after modification:")
print(reshaped_arr_modify)
print("Original array after modification:")
print(arr)
```

```python
#Arange Function
import numpy as np
b=np.arange(3)
print(b)
# Create an array using np.arange with start and stop values
c = np.arange(2, 8)
print(c)
# Create an array using np.arange with start, stop, and step values
d = np.arange(1, 10, 2)
print(d)
```

```python
#Transpose Function
import numpy as np
# Create a sample matrix
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
print("Original Matrix:")
print(matrix)

# Get the transpose using the .T attribute
transpose_matrix_T = matrix.T
print("\nTranspose using .T:")
print(transpose_matrix_T)

# Get the transpose using np.transpose()
transpose_matrix_func = np.transpose(matrix)
print("\nTranspose using np.transpose():")
print(transpose_matrix_func)

# Example with a higher-dimensional array
arr_3d = np.arange(12).reshape((2, 3, 2))
print("\nOriginal 3D array:")
print(arr_3d)

# Transpose the 3D array (swaps axes 0 and 1 by default)
transpose_arr_3d = arr_3d.T
print("\nTranspose of 3D array (.T):")
print(transpose_arr_3d)

# Transpose the 3D array by specifying axis order
transpose_arr_3d_axes = np.transpose(arr_3d, axes=(1, 0, 2))
print("\nTranspose of 3D array (np.transpose with axes=(1, 0, 2)):")
print(transpose_arr_3d_axes)
```
